# 🚀 How to Deploy a Streamlit App Using GitHub
Deploying means making your app live on the internet so other people can access it using a link.

You first upload your Streamlit app code (like space_explorer.py) to GitHub. GitHub stores your project online. Then, Streamlit Community Cloud connects to your GitHub repository, reads your code, installs the required libraries, and runs your app on its servers. Once deployed, it generates a public URL that anyone can open in their browser to use your app.

So basically:
Local app → Upload to GitHub → Deploy via Streamlit Cloud → Get live link. 🚀

## What Happens When You Deploy Your Streamlit Project?
When you deploy your Streamlit app using GitHub:
1. **Streamlit reads your code** from your GitHub repository.
2. **It installs all the libraries** listed in your `requirements.txt` file.
3. **Streamlit builds a server** just for your app.
4. **Your app goes live on the internet** — anyone can open it with a link!
5. If you update your GitHub code, Streamlit will **auto-update** your app.
